```python
import sys
from datetime import datetime, timedelta

import numpy as np
import pandas as pd
import yfinance as yf
import cvxpy as cp


def download_prices(symbols, start_date=None, end_date=None):
    """Download adjusted close prices for the given symbols."""
    if start_date is None:
        start_date = datetime.today() - timedelta(days=365 * 3)
    if end_date is None:
        end_date = datetime.today()

    data = yf.download(
        tickers=symbols,
        start=start_date.strftime("%Y-%m-%d"),
        end=end_date.strftime("%Y-%m-%d"),
        progress=False,
        auto_adjust=True,   # use adjusted close
    )
    prices = data["Close"] if isinstance(data.columns, pd.MultiIndex) else data
    return prices


def compute_statistics(prices):
    """Return daily mean returns (mu) and covariance matrix (sigma)."""
    log_ret = np.log(prices / prices.shift(1)).dropna()
    mu = log_ret.mean().values          # expected return per day
    sigma = log_ret.cov().values        # daily covariance
    return mu, sigma


def optimise_portfolio(mu, sigma, target_daily_return):
    """
    Solve:  min w.T @ Sigma @ w
            s.t. sum(w) == 1
                 mu.T @ w >= target_daily_return
                 0 <= w_i <= 1

    Returns optimal weights vector.
    """
    n = len(mu)
    w = cp.Variable(n)

    objective = cp.Minimize(cp.quad_form(w, sigma))

    constraints = [
        cp.sum(w) == 1,
        mu @ w >= target_daily_return,
        w >= 0,
        w <= 1
    ]

    prob = cp.Problem(objective, constraints)

    # Try OSQP first; fall back to ECOS → SCS if needed.
    for solver in [cp.OSQP, cp.ECOS, cp.SCS]:
        try:
            prob.solve(solver=solver)
            if w.value is not None:
                return np.array(w.value).flatten()
        except Exception as e:  # pragma: no cover
            pass

    raise RuntimeError("CVXPY failed to find a solution with any solver.")


def main(target_return=None):
    symbols = ["AAPL", "GOOGL", "AMZN", "META"]

    print(f"Downloading price data for {', '.join(symbols)} …")
    prices = download_prices(symbols)

    mu, sigma = compute_statistics(prices)

    # ------------------------------------------------------------------
    # 1. Convert the target return to a *daily* figure
    # ------------------------------------------------------------------
    if target_return is None:
        # Default: use the mean of the individual stocks as a simple target
        target_annualised = np.mean(mu) * 252
    else:
        target_annualised = float(target_return)

    target_daily = target_annualised / 252

    print(f"\nTarget annualised return set to {target_annualised:.2%} "
          f"→ daily target: {target_daily:.4f}")

    # ------------------------------------------------------------------
    # 2. Optimise
    # ------------------------------------------------------------------
    try:
        w_opt = optimise_portfolio(mu, sigma, target_daily)
    except RuntimeError as exc:
        print(f"\n❌ {exc}")
        sys.exit(1)

    # Compute portfolio metrics
    port_ret_daily = mu @ w_opt
    port_std_daily = np.sqrt(w_opt.T @ sigma @ w_opt)
    port_ret_annual = port_ret_daily * 252
    port_std_annual = port_std_daily * np.sqrt(252)

    print("\nOptimal weights:")
    for sym, weight in zip(symbols, w_opt):
        print(f"  {sym:5} : {weight:.4f}")

    print(f"\nPortfolio expected annual return : {port_ret_annual:.2%}")
    print(f"Portfolio annualised volatility   : {port_std_annual:.2%}")

main(target_return=0.12)
```
